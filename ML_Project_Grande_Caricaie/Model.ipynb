```python
import pandas as pd
from sklearn.model_selection import GroupKFold

df = pd.read_csv('data/GC_training_table_2011.csv')

CLASS_REMAP = {1: 1, 2: 2, 3: 3, 4: 4, 5: 4, 6: 5}
CLASS_NAMES = {1: 'Eaux libres', 2: 'Rivages et lieux humides', 3: 'Pelouses et prairies',
               4: 'Anthropogenic land', 5: 'Forêts'}
df['class'] = df['class_numeric'].astype(int).map(CLASS_REMAP)

PREDICTOR_COLS = ['blue', 'green', 'red', 'nir', 'swir1', 'swir2',
                   'NDVI', 'NDMI', 'MNDWI', 'NBR',
                   'elev_rel_lake', 'slope', 'dist_to_water']
assert df[PREDICTOR_COLS].isnull().sum().sum() == 0

outer_cv = GroupKFold(n_splits=7)
outer_folds = list(outer_cv.split(df[PREDICTOR_COLS], df['class'], groups=df['reserve_id']))

print(f"{df.shape[0]} rows | {len(PREDICTOR_COLS)} predictors | {len(outer_folds)} outer folds")
print(df['class'].map(CLASS_NAMES).value_counts())
```
